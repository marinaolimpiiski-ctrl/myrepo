## Цель: восстановить порядок ячеек и проверить notebook после перезапуска.

## Эскперимент: создание кода, где ячейки будут в обычном порядке, потом в ошибочном порядке.

## Тесктовая карточка, мини-эскперимент.

```python
text = "september"
print (text)
```

## пояснение
в кодовой ячейке все верно, выводится ее содержимое

## Обычный случай


```python
project_title = "Corpus Mini Lab"
print(project_title)
```

## Вывод
В кодовой ячейке верный порядок. Команда print написана после создания переменной, содержимое выводится.

## Ошибочный случай


```python
print(project_title)
project_title = "Corpus Mini Lab"
```

## Вывод
Пременная не выводится, выдает ошибку, тк ячейки расположены в неправильном порядке, т.е. сначала вводится переменная, потом ее выведение.  

## Вывод о состоянии
В ошибочном случае print(project_title) была запущена до ввода переменной, поэтому выскакивает ошибка, несоблюдение порядка ячеек
